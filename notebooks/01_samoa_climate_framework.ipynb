# Samoa climate framework — analysis notebook

This notebook is a reproducible starting point, not a forecasting model. See `../docs/evidence-manifest.md` before interpretation.

```python
import json
from pathlib import Path
import pandas as pd

data = json.loads((Path('..') / 'public/data/samoa-climate.json').read_text())
rainfall = pd.DataFrame(data['rainfall'], columns=['year', 'rainfall_anomaly_mm', 'standard_error_mm'])
sst = pd.DataFrame(data['sst'], columns=['year', 'sst_anomaly_c', 'standard_error_c'])
climate = rainfall.merge(sst, on='year', how='outer')
climate.head()
```
